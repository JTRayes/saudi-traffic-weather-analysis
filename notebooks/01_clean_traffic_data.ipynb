```python
data.dropna(inplace=
True
)
print
(
"Dataset Shape after dropping missing values:"
, data.shape)
```

```python
data.drop_duplicates(inplace=True)
print("Dataset Shape after dropping duplicates:", data.shape)
```

```python
# Get columns summary statistics
data.describe()
```

```python
# Display rows with negative values in either of the two specified columns
negative_rows = data[(data['TrafficIndexLive'] < 0) | (data['TrafficIndexWeekAgo'] < 0)]
# Display the number of rows found and show a preview
print(f"Total Rows with Negative Values: {negative_rows.shape[0]}")
print("Sample of Rows with Negative Values:")
negative_rows.style.set_properties(**{'background-color': 'white', 'color': 'red'})
```

```python
# Drop rows with negative values
data = data[(data[
'TrafficIndexLive'
] >=
0
) & (data[
'TrafficIndexWeekAgo'
] >=
0
)]
data.shape
```

```python
# Drop rows where TrafficIndexLive is above a reasonable threshold (e.g., > 60)
data = data[data[
'TrafficIndexLive'
] <=
60
]
# Verify by checking the new maximum value
data['TrafficIndexLive'].max()
```

```python
data[
'Datetime'
] = pd.to_datetime(data[
'Datetime'
], format=
'%m/%d/%y %H:%M'
, errors=
'coerce'
)
data.info()
```

```python
# Feature Engineering (if necessary)
# Create additional features if required for analysis, like 'Hour' from 'Datetime'
data[
'Year'
] = data[
'Datetime'
].dt.year
data[
'Month'
] = data[
'Datetime'
].dt.month_name()
data[
'MonthNumber'
] = data[
'Datetime'
].dt.month
data[
'Hour'
] = data[
'Datetime'
].dt.hour
data.head()
```

```python
# Create time of day column
# Early Morning:12am-6am, Morning: 6am-12pm, Afternoon: 12pm-6pm, Evening:6pm-12am
bins= np.linspace(data['Hour'].min(), data['Hour'].max(), 5)
labels = ['Early Morning', 'Morning', 'Afternoon', 'Evening']
data['Time of Day'] = pd.cut(data['Hour'], bins=bins, labels=labels, include_lowest=True)
data.head()
```

```python
if 'City' in data.columns and 'Datetime' in data.columns:
    data = data.drop_duplicates(subset=['City', 'Datetime'])
```

```python
data.columns = (
    data.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("-", "_")
    .str.lower()
)
```

```python
str_cols = data.select_dtypes(include='object').columns

for col in str_cols:
    data[col] = data[col].astype(str).str.strip().str.title()
```

```python
data = data.dropna(subset=['datetime'])
```

```python
num_cols = [
    'trafficindexlive', 'trafficindexweekago', 'jamscount',
    'jamsdelay', 'jamslength', 'traveltimelive', 'traveltimehistoric'
]

for col in num_cols:
    if col in data.columns:
        data[col] = (
            data[col]
            .astype(str)
            .str.replace(",", "")
            .str.replace("mins", "")
            .str.replace("min", "")
            .str.strip()
        )
        data[col] = pd.to_numeric(data[col], errors='coerce')

data = data.dropna(subset=num_cols)
```

```python
if 'daytype' in data.columns:
    data['daytype'] = data['daytype'].replace({
        'Weekend': 'Weekend',
        'Week End': 'Weekend',
        'Weekday': 'Weekday',
        'Week Day': 'Weekday'
    })
```

```python
output_path = 'cleaned_tarffic_data.csv'
data.to_csv(output_path, index=False)
print(f"Cleaned data saved to {output_path}")
```
